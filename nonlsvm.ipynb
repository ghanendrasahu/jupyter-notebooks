```python
from sklearn.svm import LinearSVC,SVC
x=df.iloc[:,0:5]
y=df.Species
lin_svc=LinearSVC().fit(x,y)#mul lin svm#might throw error so follow below method
svc_lin=SVC(kernel='linear').fit(x,y)
rbf_svc=SVC(kernel='rbf').fit(x,y)#radial bias function
poly_svc3=SVC(kernel='poly',degree=3).fit(x,y)#polynomial
poly_svc4=SVC(kernel='poly',degree=4).fit(x,y)
```
