```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("apples_and_oranges.csv")
```

```python

df.info()
df.describe()
```

```python
df.shape
```

```python
x=df[["Weight","Size"]]
y=df.Class
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(x,y)
```

```python
ypred=clf.predict(x)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y,ypred)
```

```python
ypred
```

```python
from sklearn.tree import plot_tree
```

```python
plot_tree(clf,feature_names=["Weight","Size"],class_names=["orange","apple"])
```
