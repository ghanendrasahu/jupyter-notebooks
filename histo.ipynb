```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
continuous_data=np.random.normal(0,1,1000)
discrete_data=np.random.poisson(5,1000)
plt.figure(figsize=(12,4))
```

```python
plt.subplot(1,2,1)
sns.histplot(continuous_data,kde=True,color="blue",bins=30)
plt.subplot(1,2,2)
sns.histplot(discrete_data,kde=True,color="green",bins=30)
plt.show()
```
