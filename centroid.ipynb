```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
np.random.seed(42)
data=np.random.normal(loc=50,scale=15,size=1000)
```

```python
range_value=np.ptp(data)
variance=np.var(data)
std_dev=np.std(data)
iqr_value=np.percentile(data,75)-np.percentile(data,25)
```

```python
mean_value=np.mean(data)
median_value=np.median(data)
mode_value=stats.mode(data)
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data,kde=True,color="lightgreen",bins=30)
plt.axvline(mean_value,color="red",linestyle='--',label=f'mean:{mean_value:.2f}')
plt.axvline(median_value,color="blue",linestyle='--',label=f'median:{median_value:}')
plt.show()
```

```python
plt.title("Data with mean,median,mod")
plt.xlabel("value")
plt.ylabel("frequency")
plt.legend()
plt.show()
```

```python
normal=np.random.normal(loc=0,scale=1,size=1000)
unifrom=np.random.uniform(low=0,high=10,size=1000)
exponential=np.random.exponential(scale=1,size=1000)
binomial=np.random.binomial(n=10,p=0.5,size=1000)
```

```python
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
sns.histplot(normal,kde=True,color="blue",bins=30)
plt.subplot(2,2,2)
sns.histplot(unifrom,kde=True,color="green",bins=30)
plt.subplot(2,2,3)
sns.histplot(exponential,kde=True,color="red",bins=30)
plt.subplot(2,2,4)
sns.histplot(binomial,kde=True,color="purple",bins=30)
plt.tight_layout()
plt.show()
```
